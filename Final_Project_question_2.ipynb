```python
# Question 2 comparing models

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, LSTM, GRU, Dense, Bidirectional
from tensorflow.keras.callbacks import History

# -----------------------------------------------------------------------------
# 0. Model‐builder functions inlined
# -----------------------------------------------------------------------------
def build_vanilla_rnn(input_shape):
    model = Sequential()
    model.add(SimpleRNN(50, activation='tanh', input_shape=input_shape))
    model.add(Dense(1))
    model.compile(optimizer='adam', loss='mse')
    return model

def build_lstm(input_shape):
    model = Sequential()
    model.add(LSTM(50, activation='tanh', input_shape=input_shape))
    model.add(Dense(1))
    model.compile(optimizer='adam', loss='mse')
    return model

def build_gru(input_shape):
    model = Sequential()
    model.add(GRU(50, activation='tanh', input_shape=input_shape))
    model.add(Dense(1))
    model.compile(optimizer='adam', loss='mse')
    return model

def build_bidirectional_rnn(input_shape):
    model = Sequential()
    model.add(Bidirectional(SimpleRNN(50, activation='tanh'), input_shape=input_shape))
    model.add(Dense(1))
    model.compile(optimizer='adam', loss='mse')
    return model

def build_bidirectional_gru(input_shape):
    model = Sequential()
    model.add(Bidirectional(GRU(50, activation='tanh'), input_shape=input_shape))
    model.add(Dense(1))
    model.compile(optimizer='adam', loss='mse')
    return model

# -----------------------------------------------------------------------------
# 1. Data preparation: sliding windows
# -----------------------------------------------------------------------------
def create_sequences(data, window_size):
    X, y = [], []
    for i in range(len(data) - window_size):
        X.append(data[i : i + window_size])
        y.append(data[i + window_size])
    return np.array(X), np.array(y)

# -----------------------------------------------------------------------------
# 2. Experiment runner
# -----------------------------------------------------------------------------
def run_experiment(prices,
                   model_builders,
                   window_sizes=(5, 20, 60),
                   epochs=50,
                   batch_size=32,
                   test_split=0.2):
    scaler = MinMaxScaler()
    scaled = scaler.fit_transform(prices.reshape(-1, 1)).flatten()
    records = []

    for win in window_sizes:
        X, y = create_sequences(scaled, win)
        split = int(len(X) * (1 - test_split))
        X_train, X_test = X[:split], X[split:]
        y_train, y_test = y[:split], y[split:]

        X_train = X_train[..., np.newaxis]
        X_test  = X_test[..., np.newaxis]

        for name, builder in model_builders.items():
            print(f"\nTraining {name} with window={win}...")
            model = builder(input_shape=(win, 1))
            history: History = model.fit(
                X_train, y_train,
                validation_data=(X_test, y_test),
                epochs=epochs,
                batch_size=batch_size,
                verbose=0
            )

            preds = model.predict(X_test, batch_size=batch_size).flatten()
            mae  = mean_absolute_error(y_test, preds)
            rmse = np.sqrt(mean_squared_error(y_test, preds))
            records.append({'Model': name, 'Window': win, 'MAE': mae, 'RMSE': rmse})

            plt.figure(figsize=(6,4))
            plt.plot(history.history['loss'],     label='train')
            plt.plot(history.history['val_loss'], label='val')
            plt.title(f'{name} (window={win}) Loss')
            plt.xlabel('Epoch'); plt.ylabel('MSE Loss')
            plt.legend(); plt.tight_layout()
            plt.show()

    return pd.DataFrame(records)

# -----------------------------------------------------------------------------
# 3. Main entrypoint
# -----------------------------------------------------------------------------
if __name__ == '__main__':
    # 3a. Load your data from CSV
    df = pd.read_csv('sp500_historical_data.csv', parse_dates=['Date'])
    df = df.sort_values('Date')
    prices = df['Close'].values.astype(float)

    # 3b. Map model names to builder functions
    model_builders = {
        'VanillaRNN': build_vanilla_rnn,
        'LSTM':       build_lstm,
        'GRU':        build_gru,
        'BiRNN':      build_bidirectional_rnn,
        'BiGRU':      build_bidirectional_gru,
    }

    # 3c. Run the experiment
    df_metrics = run_experiment(
        prices,
        model_builders,
        window_sizes=(5, 20, 60),
        epochs=50,
        batch_size=32,
        test_split=0.2
    )

    # 3d. Show pivot of results
    pivot = df_metrics.pivot_table(
        index='Window',
        columns='Model',
        values=['MAE','RMSE']
    )
    print("\n=== MAE and RMSE by Window & Model ===")
    print(pivot)
```
